```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Dataset 1-1.csv")
df
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
df
```

```python
df["ocean_proximity"]=df["ocean_proximity"].astype("category")
df["ocean_proximity"]=df["ocean_proximity"].cat.codes
```

```python
df
```

```python
x=df.drop("median_house_value",axis=1)
y=df["median_house_value"]
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.30,random_state=1)
x_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
c=model.intercept_
c
```

```python
m=model.coef_
m
```

```python
y_train[:5]
```

```python
y_train=np.array(y_train).reshape(-1)
y_train[:5]
```

```python
y_pred_train=model.predict(x_train)
y_pred_train[:5]
```

```python
from sklearn.metrics import r2_score
r2_score(y_train,y_pred_train)
```

```python
x_test[:5]
```

```python
y_test[:5]
```

```python
y_test=np.array(y_test).reshape(-1)
y_test[:5]
```

```python
y_pred_test=model.predict(x_test)
y_pred_test[:5]
```

```python
r2_score(y_test,y_pred_test)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(y_test[:50],y_pred_test[:50])
plt.xlabel("Actual Value")
plt.ylabel("Predicted Value")
plt.show()
```

```python
from sklearn.decomposition import PCA
```

```python
x[:5]
```

```python
x_scaled=scaler.fit_transform(x)
```

```python
x_scaled[:5]
```

```python
pca=PCA(n_components=8)
x_pca=pca.fit_transform(x_scaled)
x_pca.shape
```

```python
x_pca[:5]
```

```python
x_train_pca,x_test,y_train,y_test=train_test_split(x_pca,y,test_size=0.3,random_state=1)
len(x_train_pca)
```

```python
model=LinearRegression()
model.fit(x_train_pca,y_train)
```

```python
y_train=np.array(y_train).reshape(-1)
y_train[:5]
```

```python
y_pred_train_pca=model.predict(x_train_pca)
y_pred_train_pca[:5]
```

```python
r2_score(y_train,y_pred_train_pca)
```
