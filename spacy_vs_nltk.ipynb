```python
import spacy

nlp=spacy.load("en_core_web_sm")
```

```python
doc=nlp("Dr. kalam was the former president of India. Jawahar lal nehru was the first prime minister of India.")
```

```python
for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
