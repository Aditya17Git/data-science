```python
from langchain_openai import OpenAI

llm=OpenAI(temperature=0.6)
name=llm("i want to open a restaurant for Indian food. suggest a fancy name for this")
print(name)
```

```python
from langchain.prompts import PromptTemplate

prompt_template_restaurant=PromptTemplate(input_variables=['cuisine'],
                                    template="i want to open a restaurant for {cuisine}. suggest a fancy name for this"
                                   )

prompt_template_restaurant.format(cuisine="italy")
```

```python
from langchain.chains import LLMChain

restaurant_chain=LLMChain(llm=llm, prompt=prompt_template_restaurant)
restaurant_chain.run("arabic")
```

```python
prompt_template_restaurant=PromptTemplate(input_variables=['cuisine'],
                                    template="i want to open a restaurant for {cuisine}. suggest a fancy name for this"
                                   )
restaurant_chain=LLMChain(llm=llm, prompt=prompt_template_restaurant,output_key="restaurant_name")



prompt_template_items=PromptTemplate(input_variables=['restaurant_name'],
                                    template="suggest some menu items for {restaurant_name}.return it as a comma separated list"
                                   )
menu_items_chain=LLMChain(llm=llm, prompt=prompt_template_items,output_key="menu_items")
```

```python
from langchain.chains import SequentialChain

chain=SequentialChain(chains=[restaurant_chain,menu_items_chain],
                      input_variables=['cuisine'],
                      output_variables=["restaurant_name","menu_items"]
                     )
chain({'cuisine':'indian'})
```
