```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
nlp.pipeline
```

```python
doc=nlp("aditya bought a car of $15000 from 'honda motors corps'. he gifted the car to his father")
for token in doc:
    print(token,"|",token.pos_,"|",token.lemma_)
```

```python
doc=nlp("Reliance Ltd will buy Bharti Airtel for $500 billion")

for ent in doc.ents:
    print(ent.text,"|",ent.label_,"|",spacy.explain(ent.label_))
```

```python
from spacy import displacy

displacy.render(doc,style="ent")
```
