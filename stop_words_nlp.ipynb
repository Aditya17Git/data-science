```python
import spacy
```

```python
from spacy.lang.en.stop_words import STOP_WORDS
```

```python
STOP_WORDS
```

```python
nlp=spacy.load("en_core_web_sm")
doc= nlp("we our trying our level best, hope we will get the desired result soon")
for token in doc:
    if token.is_stop:
        print(token)
```

```python
# preprocessing the text with no stop words
```

```python
def preprocess(text):
    doc=nlp(text)
    no_stop_words=[token.text for token in doc if not token.is_stop and not token.is_punct]
    return " ".join(no_stop_words)
```

```python
preprocess("we our trying our level best, hope we will get the desired result soon")
```
