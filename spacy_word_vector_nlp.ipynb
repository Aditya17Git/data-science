```python
import spacy
```

```python
nlp=spacy.load("en_core_web_lg")
# write the command "python -m spacy download en_core_web_lg" on command prompt to install en_core_web_lg
```

```python
doc=nlp("bird cat dog bread skghieh")

for token in doc:
    print(token.text,'|' , "vector:",token.has_vector,'|',"out of vocab:",token.is_oov)
```

```python
base_token=nlp("tiger")
```

```python
doc=nlp("rice cat dog pizza wild bike sun jupyter tiger")

for token in doc:
    print(token.text, "|", base_token.text,"|", token.similarity(base_token))
```

```python
def similarity_score(base_word,words_to_compare):
    base_token=nlp(base_word)
    doc=nlp(words_to_compare)
    for token in doc:
         print(token.text, "|", base_token.text,"|", token.similarity(base_token))
```

```python
similarity_score("wood","pen car table burger iron enjoy chair pencil furniture")
```
