```python
import pandas as pd
```

```python
df=pd.read_csv("Fake_Real_Data.csv")
df.head()
```

```python
df.shape
```

```python
df.label.value_counts()
# to check any class imbalance
```

```python
df['new_label']=df['label'].map({"Fake":0,"Real":1})
df.head()
```

```python
import spacy

nlp=spacy.load("en_core_web_lg")
```

```python
df["vector"]=df['Text'].apply(lambda text:nlp(text).vector)
df.head()
# it takes atleast 15 minutes to run
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(df.vector.values,df.new_label,test_size=.20,random_state=1)
```

```python
x_train.shape
```

```python
x_train[:2]
```

```python
import numpy as np
```

```python
x_train=np.stack(x_train)
x_train[:2]
# to convert it into 2d array
```

```python
x_test=np.stack(x_test)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.naive_bayes import MultinomialNB

model=MultinomialNB()
model.fit(x_train,y_train)
```

```python
y_test[:5]
```

```python
y_pred=model.predict(x_test)
y_pred[:5]
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```
