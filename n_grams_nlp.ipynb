```python
# example on how to do preprocessing and then vectorization using countvector(ngram)
```

```python
corpus=["Ambani went to market and bought some fruits to feed the pets"]
```

```python
 import spacy

 # load english language model and create nlp object from it
 nlp=spacy.load("en_core_web_sm")


 def preprocess(text):
     doc=nlp(text)

     filtered_tokens=[]

     for token in doc:
         if token.is_stop or token.is_punct:
             continue
# continue keyword is used to end the current iteration in the loop and continue the next iteration
         filtered_tokens.append(token.lemma_)   
     return " ".join(filtered_tokens)
    
 preprocess("Ambani went to market and bought some fruits to feed the pets")            
```

```python
corpus_processed=[preprocess(text) for text in corpus]
corpus_processed
```

```python
v=CountVectorizer(ngram_range=(3,3))
v.fit(corpus_processed)
v.vocabulary_
```

```python
v.transform(corpus_processed).toarray()
```
